```python
df.isna().sum()
```

```python
cols_clave = ['Animal type', 'Country', 'Weight kg', 'Body Length cm', 'Gender', 'Latitude', 'Longitude']
duplicados = df[df.duplicated(subset=cols_clave, keep=False)]
duplicados.sort_values(by=cols_clave)
```

```python
#Elimina los registros duplicados.
df = df.drop_duplicates(subset=cols_clave, keep='first')
```

```python
df.duplicated().sum()
```

```python
for i in df.columns:
  error = df[i].isin(['ERROR', 'UNKNOWN']).sum()
  if error > 0:
    print(f'La columna {i} tiene {error} valores inválidos')
```

```python
df['Observation date'] = pd.to_datetime(df['Observation date'],format='%d.%m.%y', errors='coerce')
df.info()
```

```python
df["Gender"].unique()
```

```python
df["Animal code"].fillna("Desconocida")
df["Animal name"].fillna("Desconocida")
df["Gender"].fillna("not determinated")
```

```python
df.describe()
```

```python
df["Weight kg"].fillna(df["Weight kg"].median())
df["Body Length cm"].fillna(df["Body Length cm"].median())
df["Latitude"].fillna(df["Latitude"].median())
df["Longitude"].fillna(df["Longitude"].median())
```

```python
df.isna().sum()
```

```python
df['Animal type'].unique()
```

```python
reemplazos_A = {
    'European bison™': 'European bisson',
    'European bison': 'European bisson',
    'European buster': 'European bisson',
    'lynx?': 'lynx',
    'red squirel': 'red squirrel',
    'red squirrell': 'red squirrel',
    'ledgehod': 'hedgehog',
    'wedgehod': 'hedgehog',
}

df['Animal type'] = df['Animal type'].replace(reemplazos_A)
```

```python
df['Animal type'].unique()
```

```python
df['Country'].unique()
```

```python
reemplazos_B = {
    'PL': 'Poland',
    'HU': 'Hungary',
    'Hungry': 'Hungary',
    'Czech': 'Czech Republic',
    'CZ': 'Czech Republic',
}

df['Country'] = df['Country'].replace(reemplazos_B)
```

```python
df['Country'].unique()
```
